```python
from pathlib import Path
import sys

def localizar_raiz() -> Path:
    atual = Path.cwd().resolve()
    for candidato in [atual, *atual.parents]:
        if (candidato / "03-classificacao-textos").exists() and (candidato / "dados").exists():
            return candidato
    raise FileNotFoundError("Não foi possível localizar a raiz do repositório.")

ROOT = localizar_raiz()
MODULO = ROOT / "03-classificacao-textos"
sys.path.insert(0, str(MODULO))

from src.classificacao import avaliar_dataset
```
